# Обучение классических моделей

```python
import pandas as pd
import sklearn as sk
```

Прочитаем данные с отмасштабированными признаками

```python
X_train_scaled = pd.read_csv('data/X_train_scaled.csv')
X_test_scaled = pd.read_csv('data/X_test_scaled.csv')
y_train = pd.read_csv('data/y_train.csv').values.flatten() # так как методы fit принимают одномерный массив, а у нас здесь df
y_test = pd.read_csv('data/y_test.csv').values.flatten()
```

```python
X_train_scaled.head(3)
```

# BaseLine (const)

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score
```

```python
Dummy_clf = DummyClassifier(strategy='most_frequent', random_state=52)
```

```python
Dummy_clf.fit(X_train_scaled, y_train)
```

```python
y_pred = Dummy_clf.predict(X_test_scaled)
```

Посчитаем метрики для нашей baseline модели, целевая метрика - f1-macro, но будем считать ещё accuracy, precision и recall для анализа предсказаний модели, хуже от этого точно не будет

```python
f1 = f1_score(y_test, y_pred, average='macro', zero_division=0)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
```

```python
metrics_const = pd.DataFrame(
    columns=['f1-macro', 'accuracy', 'precision', 'recall'],
    data=[[f1, accuracy, precision, recall]],
    index=['Const']
)
```

```python
metrics_const
```

# Random Forest

Тут будем использовать данные без масштабирования числовых признаков

```python
X_train_encoded = pd.read_csv('data/X_train_encoded.csv')
X_test_encoded = pd.read_csv('data/X_test_encoded.csv')
```

```python
metrics_rf
```

# CatBoost

Для catboost будем использовать заранее подготовленные данные без кодирования признаков и масштабирования, catboost сам отлично сделает это

```python
X_train_raw = pd.read_csv('data/X_train_raw.csv')
X_test_raw = pd.read_csv('data/X_test_raw.csv')
```

```python
cat_cols = X_train_raw.select_dtypes(include='str').columns.tolist()
```

```python
import catboost
```

```python
model = catboost.CatBoostClassifier(random_state=52, class_weights=[1, 3], verbose=0, learning_rate=0.1, depth=6, iterations=1000)
```

```python
model.fit(X_train_raw, y_train, cat_features=cat_cols)
```

```python
y_pred = model.predict(X_test_raw)
```

```python
f1_cat = f1_score(y_test, y_pred, average='macro')
accuracy_cat = accuracy_score(y_test, y_pred)
precision_cat = precision_score(y_test, y_pred)
recall_cat = recall_score(y_test, y_pred)
```

```python
metrics_cat = pd.DataFrame(
    columns=['f1-macro', 'accuracy', 'precision', 'recall'],
    data=[[f1_cat, accuracy_cat, precision_cat, recall_cat]],
    index=['CatBoost']
)
```

```python
metrics_cat
```
